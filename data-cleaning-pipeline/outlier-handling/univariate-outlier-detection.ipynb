```python
import numpy as np
```

```python
n_samples = 5000
n_outliers = 10

n_inliers = n_samples - n_outliers
```

```python
X_in = np.random.randn(n_inliers)
X_out = np.concatenate([100 + np.random.randn(int(n_outliers/2)), -10 + np.random.randn(int(n_outliers/2))])
X = np.concatenate([X_in, X_out])

print(np.sort(X_out))
```

### Univariate Outlier Detection
- Z-Score (suitable for normal, sensitive to outliers)
- Robust Z-Score (suitable for normal, resistant to outliers)
- IQR 
- Winsorization

```python
# Z-Score
mean = X.mean()
std = X.std()

z_score = (X - mean) / std
outliers = X[np.abs(z_score) >= 3].copy()
outliers = np.sort(outliers)

print(outliers)
```

```python
# Modified Z-Score
median = np.median(X)
mad = np.median(np.abs(X - median))

z_score_modified = 0.6745 * (X - median) / mad
outliers = X[np.abs(z_score_modified) >= 3].copy()
outliers = np.sort(outliers)

print(outliers)
```

```python
# IQR
q1 = np.quantile(X, 0.25)
q3 = np.quantile(X, 0.75)
iqr = q3 - q1

outliers = X[(X <= q1 - 1.5 * iqr) | (X >= q3 + 1.5 * iqr)].copy()
outliers = np.sort(outliers)

print(outliers)
```

```python
# Winsorization
from scipy.stats.mstats import winsorize
inliers = winsorize(X, limits=[0.025, 0.025])

outliers = X[(X <= np.quantile(X, 0.025)) | (X >= np.quantile(X, 0.975))]
outliers = np.sort(outliers)

print(outliers)
```

**Conclusion:**
- Z-Score and Robust Z-Score methods good for detecting global outliers
- Robust Z-Score method is resistant to outliers
- Quantile based methods detect a lot of local outliers in addition to global ones
