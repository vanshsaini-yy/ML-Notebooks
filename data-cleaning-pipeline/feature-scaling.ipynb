```python
import pandas as pd
import numpy as np
```

```python
n_samples = 5000
outlier_fraction = 0.05

n_outliers = int(n_samples * outlier_fraction)
n_inliers = n_samples - n_outliers

data = pd.DataFrame({
    'X' : np.concat([np.random.randn(n_inliers), 10 + 5 * np.random.randn(n_outliers)]),
    'Y' : np.random.choice([-2.0, -1.0, 0.0, 1.0, 2.0], size=n_samples, p = [0.2, 0.4, 0.1, 0.2, 0.1])
})

zero_count = (data['Y'] == 0.0).sum()
zero_count
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler, MaxAbsScaler

temp_data = data.copy()
scaler = MinMaxScaler()
temp_data.loc[:] = scaler.fit_transform(temp_data)
display(temp_data['X'].describe())

zero_count = (temp_data['Y'] == 0.0).sum()
display(zero_count)
```

```python
temp_data = data.copy()
scaler = StandardScaler()
temp_data.loc[:] = scaler.fit_transform(temp_data)
display(temp_data['X'].describe())

zero_count = (temp_data['Y'] == 0.0).sum()
display(zero_count)
```

```python
temp_data = data.copy()
scaler = RobustScaler()
temp_data.loc[:] = scaler.fit_transform(temp_data)
display(temp_data['X'].describe())

zero_count = (temp_data['Y'] == 0.0).sum()
display(zero_count)
```

```python
temp_data = data.copy()
scaler = MaxAbsScaler()
temp_data.loc[:] = scaler.fit_transform(temp_data)
display(temp_data['X'].describe())

zero_count = (temp_data['Y'] == 0.0).sum()
display(zero_count)
```

### **Conclusion**
- MinMaxScaler and MaxAbsScaler map feature to fixed range.
- StandardScaler makes the feature zero mean and unit variance.
- RobustScaler is best when outliers are present; outliers show up as high value.
- MaxAbsScaler preserves the number of zero entries.
